### Tratamento da variável Age

A variável `Age` apresenta valores ausentes. Para evitar a perda de
observações, os valores faltantes foram substituídos pela mediana da variável.

A mediana foi escolhida por ser menos sensível a valores extremos do que a
média, característica relevante considerando a distribuição das variáveis
numéricas da base.

````python
df['Age'] = df['Age'].fillna(df['Age'].median())
````

````python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
````

A variável `Embarked` possui apenas dois valores ausentes. Como se trata de
uma variável categórica, os valores foram preenchidos utilizando a moda,
representada pela categoria mais frequente.

````python
df['CabinKnown'] = df['Cabin'].notna().astype(int)
````

````python
df['Child'] = (df['Age']<= 12).astype(int) 
````

A regra será:

1 = passageiro com até 12 anos
0 = passageiro acima de 12 anos

````python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
````

````python
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
````

````python
survival_rate = df ['Survived'].mean()
print(f"Taxa geral de sobrevivencia: {survival_rate:.2%}")
````

# Sobrevivência por sexo 

````python
survival_by_sex = df.groupby('Sex')['Survived'].mean()
print(survival_by_sex)

survival_by_sex = (
    df.groupby('Sex')['Survived']
        .agg(['count', 'mean'])
        .rename(columns={
            'count': 'Passageiros',
            'mean':'Taxa_Sobrevivencia'
        })
)
survival_by_sex['Taxa_Sobrevivencia'] *= 100
display(survival_by_sex)
````

````python
survival_by_class = (
    df.groupby('Pclass')['Survived']
    .agg(['count','mean'])
    .rename(columns={
        'count': 'Passageiros',
    'mean':'Taxa_Sobrevivencia'})
)

survival_by_class ['Taxa_Sobrevivencia'] *=100
display(survival_by_class)
````

# Regressão Logistica para prver os Sobreviventes. 

````python
from sklearn.model_selection import train_test_split

X = df[["Pclass", "Sex", "Age", "Fare"]]
y = df["Survived"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y)
````

````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

colunas_numericas = ["Pclass", "Age", "Fare"]
colunas_categoricas = ["Sex"]

preprocessamento_numerico = Pipeline([
    ("imputacao", SimpleImputer(strategy="median")),
    ("escala", StandardScaler())
])

preprocessamento_categorico = Pipeline([
    ("imputacao", SimpleImputer(strategy="most_frequent")),
    ("codificacao", OneHotEncoder(handle_unknown="ignore"))
])

preprocessamento = ColumnTransformer([
    ("numericas", preprocessamento_numerico, colunas_numericas),
    ("categoricas", preprocessamento_categorico, colunas_categoricas)
])

modelo = Pipeline([
    ("preprocessamento", preprocessamento),
    ("regressao", LogisticRegression(max_iter=1000))
])

modelo.fit(X_train, y_train)
````

````python
#gerando o modelo preditivo 
y_pred = modelo.predict(X_test)
````

````python
#Bibliotecas Necessárias.

from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

#calculando a matriz de confusão. 
cm = confusion_matrix(y_test, y_pred)

#visualizando a matriz 
plt.figure(figsize=(6,4))

sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
           xticklabels=["Não sobreviveu", "Sobreviveu"],
           yticklabels=["Não sobreviveu", "Sobreviveu"])

plt.title("Matriz de Confusão")
plt.xlabel("Prervisão de modelo")
plt.ylabel("Valor real")

plt.tight_layout()
plt.show ()
````
